````python
import random
from typing import List, Tuple

# Cursos: (nombre, número de estudiantes)
courses = [
    ('MAT', 45),  # 0
    ('FIS', 30),  # 1
    ('QUI', 35),  # 2
    ('PRG', 40),  # 3
    ('LIT', 25),  # 4
    ('HIS', 30),  # 5
    ('ING', 35),  # 6
    ('EDF', 50)   # 7
]

# Bloques de tiempo disponibles
time_blocks = [1, 2, 3, 4, 5]
block_names = {
    1: 'Lun 8-10',
    2: 'Lun 10-12',
    3: 'Mie 8-10',
    4: 'Mie 10-12',
    5: 'Vie 8-10'
}

# Salones: {nombre: capacidad}
rooms = {
    'A': 40,
    'B': 30,
    'C': 50
}

# Índices de cursos que comparten profesor (no pueden ser simultáneos)
same_professor = [
    (0, 1),  # MAT y FIS
    (2, 3)   # QUI y PRG
]

print("✓ Configuración cargada")
print(f"  Cursos: {len(courses)}")
print(f"  Bloques: {len(time_blocks)}")
print(f"  Salones: {len(rooms)}")
````

# TAREA 1: Función de Aptitud (Fitness) [40 puntos]

## Instrucciones

Completa la función `fitness()` que evalúa qué tan bueno es un horario.

### Sistema de Puntuación:

**Puntaje Base:** 200 puntos

**Penalizaciones por restricciones DURAS:**
- **-50 puntos** por cada conflicto de salón (dos cursos mismo bloque y salón)
- **-50 puntos** por cada salón con capacidad insuficiente
- **-50 puntos** por cada par de cursos del mismo profesor en mismo bloque

**Penalizaciones por restricciones BLANDAS:**
- **-5 puntos** por cada curso en bloques tardíos (bloques 4 o 5)
- **-3 puntos** si un curso pequeño (<35 estudiantes) usa salón grande (>45 capacidad)

### Pseudocódigo:

```
función fitness(individual):
    score = 200
    
    # 1. Conflictos de salones
    para cada par de cursos (i, j) donde i < j:
        si tienen mismo bloque Y mismo salón:
            score -= 50
    
    # 2. Capacidad de salones
    para cada curso i:
        si estudiantes[i] > capacidad del salón:
            score -= 50
    
    # 3. Profesores compartidos
    para cada par en same_professor:
        si ambos en mismo bloque:
            score -= 50
    
    # 4. Bloques tardíos
    para cada curso:
        si bloque >= 4:
            score -= 5
    
    # 5. Desperdicio de espacio
    para cada curso:
        si estudiantes < 35 Y capacidad > 45:
            score -= 3
    
    retornar score
```

---

````python
def fitness(individual):
    """
    Calcula la aptitud de un horario.
    
    Args:
        individual: Lista de tuplas [(bloque, salón), ...] para cada curso
    
    Returns:
        score: Puntaje de aptitud (mayor es mejor)
    """
    # Puntaje base
    score = 200
    
       
    # 1. RESTRICCIÓN DURA: Conflictos de salones (-50 puntos cada uno)
    # Penalizar si dos cursos están en el mismo bloque y salón
    # Pista: usa dos bucles anidados con range(len(individual))
    # Compara individual[i] con individual[j] donde i < j
    
    for i in range(len(individual)):
        for j in range(i + 1, len(individual)): # i+1 para no comparar un curso con el mismo
            
            bloque_i, salon_i = individual[i]
            bloque_j, salon_j = individual[j]
            
            if bloque_i == bloque_j and salon_i == salon_j: # no se pueden dictar dos clases en el mismo bloque y mismo salon 
                score -= 50
    
    
    # 2. RESTRICCIÓN DURA: Capacidad de salones (-50 puntos cada uno)
    # Penalizar si un curso tiene más estudiantes que la capacidad del salón
    # Pista: courses[i] = (nombre, estudiantes)
    #        individual[i] = (bloque, salón)
    #        rooms[salón] = capacidad
    
    for i in range(len(individual)): # verifica la cantidad de estudiantes y la capacidad del salón
        
        bloque, salon = individual[i]
        nombre_curso, estudiantes = courses[i]
        capacidad_salon = rooms[salon]
        
        if estudiantes > capacidad_salon:
            score -= 50
    
    
    # 3. RESTRICCIÓN DURA: Profesores compartidos (-50 puntos cada uno)
    # Penalizar si cursos del mismo profesor están en el mismo bloque
    # Pista: same_professor contiene tuplas (curso1_idx, curso2_idx)
    #        compara individual[curso1_idx][0] con individual[curso2_idx][0]
    
    for curso1_idx, curso2_idx in same_professor: # revisa si esta en el mismo bloque por si el profe tiene que estar en 2 salones al mismo tiempo
        
        bloque1 = individual[curso1_idx][0]
        bloque2 = individual[curso2_idx][0]
        
        if bloque1 == bloque2:
            score -= 50
    
    
    # 4. RESTRICCIÓN BLANDA: Bloques tardíos (-5 puntos cada uno)
    # Penalizar cursos en bloques 4 o 5
    # Pista: individual[i][0] es el bloque del curso i
    
    for i in range(len(individual)): # penalización para evitar los bloques de más tarde
        
        bloque = individual[i][0]
        
        if bloque >= 4:
            score -= 5
    
    
    # 5. RESTRICCIÓN BLANDA: Desperdicio de espacio (-3 puntos cada uno)
    # Penalizar si curso pequeño (<35 est.) usa salón grande (>45 cap.)
    # Pista: courses[i][1] es el número de estudiantes
    #        rooms[individual[i][1]] es la capacidad del salón
    
    
    for i in range(len(individual)): # compara la cantidad de estidoantes con la capacidad del salón
        
        bloque, salon = individual[i]
        nombre_curso, estudiantes = courses[i]
        capacidad_salon = rooms[salon]
        
        if estudiantes < 35 and capacidad_salon > 45:
            score -= 3
    
    
    return score


# Probar tu función
test_ind = create_random_individual()
print_schedule(test_ind)
print(f"\nFitness: {fitness(test_ind)}")
````

# TAREA 2: Operador de Cruce (Crossover) [30 puntos]

## Instrucciones

Implementa un operador de **cruce de un punto** que combine dos horarios padres.

### Funcionamiento:
1. Selecciona un punto de corte aleatorio (entre 1 y 7)
2. Copia los primeros cursos del padre 1 (hasta el punto de corte)
3. Copia los cursos restantes del padre 2 (después del punto de corte)

### Ejemplo:
```python
Padre 1: [(1,'A'), (2,'B'), (3,'C'), (4,'A'), (5,'B'), (1,'C'), (2,'A'), (3,'B')]
Padre 2: [(2,'C'), (3,'A'), (4,'B'), (5,'C'), (1,'A'), (2,'B'), (3,'C'), (4,'A')]
Punto: 4

Hijo:    [(1,'A'), (2,'B'), (3,'C'), (4,'A'), (1,'A'), (2,'B'), (3,'C'), (4,'A')]
         |------------- P1 -------------|------------ P2 -------------|
```

### Pseudocódigo:
```
función crossover(padre1, padre2):
    punto = número aleatorio entre 1 y 7
    hijo = padre1[0:punto] + padre2[punto:]
    retornar hijo
```

---

````python
def crossover(parent1, parent2):
    """
    Crea un hijo combinando dos padres usando cruce de un punto.
    
    Args:
        parent1: Primer padre (horario)
        parent2: Segundo padre (horario)
    
    Returns:
        child: Nuevo horario combinando ambos padres
    """
    point = random.randint(1, 7)  # Punto de corte entre 1 y 7
    child = parent1[:point] + parent2[point:] # Combina las partes de ambos padres
    return child

# Probar tu operador
p1 = create_random_individual()
p2 = create_random_individual()

print("Padre 1:")
print_schedule(p1)
print(f"Fitness: {fitness(p1)}")

print("\nPadre 2:")
print_schedule(p2)
print(f"Fitness: {fitness(p2)}")

child = crossover(p1, p2)
print("\nHijo:")
print_schedule(child)
print(f"Fitness: {fitness(child)}")
````

### Funcionamiento:

### Pseudocódigo:
```

    para cada posición i:
        si random.random() < prob:
            tipo = elegir aleatoriamente entre [1, 2, 3]

                nuevo_bloque = elegir bloque aleatorio
                individual[i] = (nuevo_bloque, salón_actual)

                individual[i] = (bloque_actual, nuevo_salón)
            sino:  # Cambiar ambos

                nuevo_salón = elegir salón aleatorio

```

---

````python
def mutate(individual, prob=0.2):
    """
    Muta un horario cambiando aleatoriamente bloques o salones.
    
    Args:
        individual: Horario a mutar
        prob: Probabilidad de mutación por curso (default: 20%)
    
    Returns:
        individual: Horario mutado
    """
    
    # Recorrer cada curso
    for i in range(len(individual)):
        # Decidir si mutar este curso (probabilidad prob)
        if random.random() < prob:
            
            # Obtener valores actuales
            current_block, current_room = individual[i]
            
            # Elegir tipo de mutación: 1=solo bloque, 2=solo salón, 3=ambos
            mutation_type = random.choice([1, 2, 3])
            
            if mutation_type == 1:
                # Mutar solo el bloque de tiempo
                new_block = random.choice(time_blocks)
                individual[i] = (new_block, current_room)
                
            elif mutation_type == 2:
                # Mutar solo el salón
                new_room = random.choice(list(rooms.keys()))
                individual[i] = (current_block, new_room)
                
            else:
                new_block = random.choice(time_blocks)
                new_room = random.choice(list(rooms.keys()))
                individual[i] = (new_block, new_room)
    
    return individual

# Probar tu operador
original = create_random_individual()
print("Original:")
print_schedule(original)
print(f"Fitness: {fitness(original)}")

mutated = [x for x in original]  # Copia
mutated = mutate(mutated, prob=0.5)  # 50% para ver más mutaciones

print("\nMutado (prob=50%):")
print_schedule(mutated)
print(f"Fitness: {fitness(mutated)}")
````

# Función de Evolución (Proporcionada)

Esta función combina todos tus operadores:

````python
def selection(population):
    """Selección por torneo de 3 individuos."""
    tournament = random.sample(population, 3)
    return max(tournament, key=fitness)

def evolve(population, generations=50, verbose=True):
    best_overall = None
    best_overall_fitness = float('-inf')
    
    fitness_history = []   # Guardamos la evolución
    
    for gen in range(generations):
        
        # Aplicamos elitismo, mantenemos los 2 mejores
        sorted_pop = sorted(population, key=fitness, reverse=True)
        new_population = sorted_pop[:2]
        
        #generamos el resto de la población
        while len(new_population) < len(population):
            parent1 = selection(population)
            parent2 = selection(population)
            child = crossover(parent1, parent2)
            child = mutate(child)
            new_population.append(child)
        
        population = new_population
        
        #buscamos el mejor de esta generación
        best = max(population, key=fitness)
        best_fit = fitness(best)
        
        fitness_history.append(best_fit)  # Guardamos el historial
        
        if best_fit > best_overall_fitness:
            best_overall = best
            best_overall_fitness = best_fit
        
        if verbose and (gen % 10 == 0 or gen == generations - 1):
            print(f"Gen {gen+1:3d}: Mejor fitness = {best_fit:6.1f}")
    
    return best_overall, fitness_history  # retornamos el mejor fitness de la generación y el historial de los mejores fitness

best_solution, history = evolve(initial_pop, generations=100)
````

````python
#Opcion 1 - Cruce de dos puntos
def crossover(parent1, parent2):
    """
    Crea un hijo combinando dos padres usando cruce de dos puntos.
    Aleatoriamente usa el patrón P1+P2+P1 o P2+P1+P2.
    
    Args:
        parent1: Primer padre (horario)
        parent2: Segundo padre (horario)
    
    Returns:
        child: Nuevo horario combinando ambos padres
    """
    point1 = random.randint(1, 6)  # Primer punto de corte
    point2 = random.randint(point1 + 1, 7)  # Segundo punto (siempre mayor que point1)

    if random.random() < 0.5:
        # Patrón P1 + P2 + P1
        child = parent1[:point1] + parent2[point1:point2] + parent1[point2:]
    else:
        # Patrón P2 + P1 + P2
        child = parent2[:point1] + parent1[point1:point2] + parent2[point2:]

    return child

# Probar tu operador
p1 = create_random_individual()
p2 = create_random_individual()

print("Padre 1:")
print_schedule(p1)
print(f"Fitness: {fitness(p1)}")

print("\nPadre 2:")
print_schedule(p2)
print(f"Fitness: {fitness(p2)}")

child = crossover(p1, p2)
print("\nHijo:")
print_schedule(child)
print(f"Fitness: {fitness(child)}")
````

````python

##Opcion 2 - Mutación inteligente

def mutate(individual, prob=0.2):
    """
    Mutación inteligente: solo corrige violaciones específicas detectadas.
    No aplica mutaciones aleatorias.
    
    Args:
        individual: Horario a mutar
        prob: Parámetro mantenido por compatibilidad (no se usa)
    
    Returns:
        individual: Horario mutado
    """
    
    # Recorrer cada curso
    for i in range(len(individual)):
        
        # Obtener valores actuales
        current_block, current_room = individual[i]
        course_students = courses[i][1]
        
        # Mutacion 1 - Dos cursos estan usando el mismo salon en el mismo bloque de tiempo
        if room_has_conflict(current_room, current_block, individual, i):
            # Buscar salones disponibles con capacidad suficiente
            available_rooms = [r for r in rooms 
                             if r != current_room 
                             and rooms[r] >= course_students
                             and not room_has_conflict(r, current_block, individual, i)]
            
            # Si no hay salones ideales, buscar al menos uno diferente con capacidad
            if not available_rooms:
                available_rooms = [r for r in rooms 
                                 if r != current_room 
                                 and rooms[r] >= course_students]
            
            if available_rooms:
                new_room = random.choice(available_rooms)
                individual[i] = (current_block, new_room)                
                        
        # Mutacion 2 - El curso tiene mas estudiantes que la capacidad del salon
        elif capacity_violation(current_room, i):
            # Buscar salones con capacidad suficiente
            suitable_rooms = [r for r in rooms if rooms[r] >= course_students]
            if suitable_rooms:
                new_room = random.choice(suitable_rooms)
                individual[i] = (current_block, new_room)
        
        # Mutacion 3 - Dos cursos comparten el mismo profesor en el mismo bloque de tiempo
        else:
            # Buscar directamente si este curso tiene conflicto de profesor
            conflicting_course = None
            conflicting_block = None
            
            for c1, c2 in same_professor:
                if c1 == i and individual[c1][0] == individual[c2][0]:
                    conflicting_course = c2
                    conflicting_block = individual[c2][0]
                    break
                elif c2 == i and individual[c1][0] == individual[c2][0]:
                    conflicting_course = c1
                    conflicting_block = individual[c1][0]
                    break
            
            # Si hay conflicto de profesor, cambiar el bloque
            if conflicting_block is not None:
                available_blocks = [b for b in time_blocks if b != conflicting_block]
                if available_blocks:
                    new_block = random.choice(available_blocks)
                    individual[i] = (new_block, current_room)
    
    return individual

def room_has_conflict(room, block, individual, i):
    """Verificar si el salón está siendo usado por otro curso en el mismo bloque."""
    for j in range(len(individual)):
        if j != i and individual[j][0] == block and individual[j][1] == room:
            return True
    return False

def capacity_violation(room, i):
    """Verificar si el curso tiene mas estudiantes que la capacidad del salon."""
    course_students = courses[i][1]
    room_capacity = rooms[room]
    return course_students > room_capacity


# Probar tu operador
original = create_random_individual()
print("Original:")
print_schedule(original)
print(f"Fitness: {fitness(original)}")

mutated = [x for x in original]  # Copia
mutated = mutate(mutated, prob=0.5)  # 50% para ver más mutaciones

print("\nMutado (prob=50%):")
print_schedule(mutated)
print(f"Fitness: {fitness(mutated)}")
````
