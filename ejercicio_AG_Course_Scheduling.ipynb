Un colegio necesita programar **8 cursos** en **5 bloques de tiempo** disponibles.

1. **Matemáticas (MAT)** - 45 estudiantes
2. **Física (FIS)** - 30 estudiantes  
3. **Química (QUI)** - 35 estudiantes
4. **Programación (PRG)** - 40 estudiantes

## Configuración Inicial

Ejecuta esta celda para cargar los datos del problema:

````python
import random
from typing import List, Tuple

# Cursos: (nombre, número de estudiantes)
courses = [
    ('MAT', 45),  # 0
    ('FIS', 30),  # 1
    ('QUI', 35),  # 2
    ('PRG', 40),  # 3
    ('LIT', 25),  # 4
    ('HIS', 30),  # 5
    ('ING', 35),  # 6
    ('EDF', 50)   # 7
]

# Bloques de tiempo disponibles
time_blocks = [1, 2, 3, 4, 5]
block_names = {
    1: 'Lun 8-10',
    2: 'Lun 10-12',
    3: 'Mie 8-10',
    4: 'Mie 10-12',
    5: 'Vie 8-10'
}

# Salones: {nombre: capacidad}
rooms = {
    'A': 40,
    'B': 30,
    'C': 50
}

# Índices de cursos que comparten profesor (no pueden ser simultáneos)
same_professor = [
    (0, 1),  # MAT y FIS
    (2, 3)   # QUI y PRG
]

print("✓ Configuración cargada")
print(f"  Cursos: {len(courses)}")
print(f"  Bloques: {len(time_blocks)}")
print(f"  Salones: {len(rooms)}")
````

## Funciones Auxiliares (Proporcionadas)

Estas funciones te ayudarán a visualizar y crear individuos:

````python
def print_schedule(individual):
    """Imprime un horario de forma legible."""
    print("\n" + "="*60)
    print("HORARIO")
    print("="*60)
    for i, (block, room) in enumerate(individual):
        name, students = courses[i]
        print(f"{name:3s} | {block_names[block]:10s} | Salón {room} (Cap:{rooms[room]:2d}) | Est:{students:2d}")
    print("="*60)

def create_random_individual():
    """Crea un individuo aleatorio."""
    individual = []
    for i in range(len(courses)):
        block = random.choice(time_blocks)
        room = random.choice(list(rooms.keys()))
        individual.append((block, room))
    return individual

def create_population(size=20):
    """Crea una población inicial."""
    return [create_random_individual() for _ in range(size)]

# Probar
sample = create_random_individual()
print_schedule(sample)
````

# TAREA 1: Función de Aptitud (Fitness) [40 puntos]

## Instrucciones

Completa la función `fitness()` que evalúa qué tan bueno es un horario.

### Sistema de Puntuación:

**Puntaje Base:** 200 puntos

**Penalizaciones por restricciones DURAS:**
- **-50 puntos** por cada conflicto de salón (dos cursos mismo bloque y salón)
- **-50 puntos** por cada salón con capacidad insuficiente
- **-50 puntos** por cada par de cursos del mismo profesor en mismo bloque

**Penalizaciones por restricciones BLANDAS:**
- **-5 puntos** por cada curso en bloques tardíos (bloques 4 o 5)
- **-3 puntos** si un curso pequeño (<35 estudiantes) usa salón grande (>45 capacidad)

### Pseudocódigo:

```
función fitness(individual):
    score = 200
    
    # 1. Conflictos de salones
    para cada par de cursos (i, j) donde i < j:
        si tienen mismo bloque Y mismo salón:
            score -= 50
    
    # 2. Capacidad de salones
    para cada curso i:
        si estudiantes[i] > capacidad del salón:
            score -= 50
    
    # 3. Profesores compartidos
    para cada par en same_professor:
        si ambos en mismo bloque:
            score -= 50
    
    # 4. Bloques tardíos
    para cada curso:
        si bloque >= 4:
            score -= 5
    
    # 5. Desperdicio de espacio
    para cada curso:
        si estudiantes < 35 Y capacidad > 45:
            score -= 3
    
    retornar score
```

---

````python
def fitness(individual):
    """
    Calcula la aptitud de un horario.
    """
    
    # Puntaje base
    score = 200
    
       
    # 1. RESTRICCIÓN DURA: Conflictos de salones (-50 puntos cada uno)
    # Penalizar si dos cursos están en el mismo bloque y salón
    
    for i in range(len(individual)):

        # i+1 para no comparar un curso consigo mismo y contar el mismo conflicto 2 veces
        for j in range(i + 1, len(individual)): 
            
            bloque_i, salon_i = individual[i]
            bloque_j, salon_j = individual[j]
            
            if bloque_i == bloque_j and salon_i == salon_j: # no se pueden dictar dos clases en el mismo bloque y mismo salon 
                score -= 50
    
    
    # 2. RESTRICCIÓN DURA: Capacidad de salones (-50 puntos cada uno)
    # Penalizar si un curso tiene más estudiantes que la capacidad del salón

    # verifica la cantidad de estudiantes y la capacidad del salón
    for i in range(len(individual)): 
        
        bloque, salon = individual[i] 
        nombre_curso, estudiantes = courses[i]
        capacidad_salon = rooms[salon]
        
        if estudiantes > capacidad_salon:
            score -= 50
    
    
    # 3. RESTRICCIÓN DURA: Profesores compartidos (-50 puntos cada uno)
    # Penalizar si cursos del mismo profesor están en el mismo bloque
  
    # revisa si esta en el mismo bloque por si el profe tiene que estar en 2 salones al mismo tiempo
    for curso1_idx, curso2_idx in same_professor: 
        
        bloque1 = individual[curso1_idx][0]
        bloque2 = individual[curso2_idx][0]
        
        if bloque1 == bloque2:
            score -= 50
    
    
    # 4. RESTRICCIÓN BLANDA: Bloques tardíos (-5 puntos cada uno)
 
    # penalización para evitar los bloques de más tarde
    for i in range(len(individual)): 
        
        bloque = individual[i][0]
        
        if bloque >= 4:
            score -= 5
    
    
    # 5. RESTRICCIÓN BLANDA: Desperdicio de espacio (-3 puntos cada uno)
    # Penalizar si curso pequeño (<35 est.) usa salón grande (>45 cap.)
    
    # compara la cantidad de estudiantes con la capacidad del salón para optimizar uso del espacio
    for i in range(len(individual)):

        bloque, salon = individual[i]
        nombre_curso, estudiantes = courses[i]
        capacidad_salon = rooms[salon]
        
        if estudiantes < 35 and capacidad_salon > 45:
            score -= 3
    
    
    return score


# Probar la función
test_ind = create_random_individual()
print_schedule(test_ind)
print(f"\nFitness: {fitness(test_ind)}")
````

# TAREA 2: Operador de Cruce (Crossover) [30 puntos]

## Instrucciones

Implementa un operador de **cruce de un punto** que combine dos horarios padres.

### Funcionamiento:
1. Selecciona un punto de corte aleatorio (entre 1 y 7)
2. Copia los primeros cursos del padre 1 (hasta el punto de corte)
3. Copia los cursos restantes del padre 2 (después del punto de corte)

### Ejemplo:
```python
Padre 1: [(1,'A'), (2,'B'), (3,'C'), (4,'A'), (5,'B'), (1,'C'), (2,'A'), (3,'B')]
Padre 2: [(2,'C'), (3,'A'), (4,'B'), (5,'C'), (1,'A'), (2,'B'), (3,'C'), (4,'A')]
Punto: 4

Hijo:    [(1,'A'), (2,'B'), (3,'C'), (4,'A'), (1,'A'), (2,'B'), (3,'C'), (4,'A')]
         |------------- P1 -------------|------------ P2 -------------|
```

### Pseudocódigo:
```
función crossover(padre1, padre2):
    punto = número aleatorio entre 1 y 7
    hijo = padre1[0:punto] + padre2[punto:]
    retornar hijo
```

---

````python
def crossover(parent1, parent2):
    """
    Crea un hijo combinando dos padres usando cruce de un punto.
    
    Args:
        parent1: Primer padre (horario)
        parent2: Segundo padre (horario)
    
    Returns:
        child: Nuevo horario combinando ambos padres
    """
    point = random.randint(1, 7)  # Punto de corte entre 1 y 7
    child = parent1[:point] + parent2[point:] # Combina las partes de ambos padres
    return child

# Probar tu operador
p1 = create_random_individual()
p2 = create_random_individual()

print("Padre 1:")
print_schedule(p1)
print(f"Fitness: {fitness(p1)}")

print("\nPadre 2:")
print_schedule(p2)
print(f"Fitness: {fitness(p2)}")

child = crossover(p1, p2)
print("\nHijo:")
print_schedule(child)
print(f"Fitness: {fitness(child)}")
````

# TAREA 3: Operador de Mutación [30 puntos]

## Instrucciones

Implementa un operador de mutación que introduzca cambios aleatorios.

### Funcionamiento:
Para cada curso, con probabilidad `prob` (20%):
- Decide aleatoriamente si cambiar:
  - Solo el bloque de tiempo
  - Solo el salón  
  - Ambos

### Ejemplo:
```python
Original: [(1,'A'), (2,'B'), (3,'C'), (4,'A'), ...]
Mutación en posición 1 (cambiar ambos):
Mutado:   [(1,'A'), (5,'C'), (3,'C'), (4,'A'), ...]
                     ^^^^^^^
```

### Pseudocódigo:
```
función mutate(individual, prob=0.2):
    para cada posición i:
        si random.random() < prob:
            tipo = elegir aleatoriamente entre [1, 2, 3]
            si tipo == 1:  # Cambiar solo bloque
                nuevo_bloque = elegir bloque aleatorio
                individual[i] = (nuevo_bloque, salón_actual)
            sino si tipo == 2:  # Cambiar solo salón
                nuevo_salón = elegir salón aleatorio
                individual[i] = (bloque_actual, nuevo_salón)
            sino:  # Cambiar ambos
                nuevo_bloque = elegir bloque aleatorio
                nuevo_salón = elegir salón aleatorio
                individual[i] = (nuevo_bloque, nuevo_salón)
    retornar individual
```

---

````python
def mutate(individual, prob = 0.2):
    """
    Muta un horario cambiando aleatoriamente bloques o salones.
    
    Args:
        individual: Horario a mutar
        prob: Probabilidad de mutación por curso (default: 20%)
    
    Returns:
        individual: Horario mutado
    """
    
    # Recorrer cada curso
    for i in range(len(individual)):
        # Decidir si mutar este curso (probabilidad prob)
        if random.random() < prob:
            
            # Obtener valores actuales
            current_block, current_room = individual[i]
            
            # Elegir tipo de mutación: 1=solo bloque, 2=solo salón, 3=ambos
            mutation_type = random.choice([1, 2, 3])
            
            if mutation_type == 1:
                # Mutar solo el bloque de tiempo
                new_block = random.choice(time_blocks)
                individual[i] = (new_block, current_room)
                
            elif mutation_type == 2:
                # Mutar solo el salón
                new_room = random.choice(list(rooms.keys()))
                individual[i] = (current_block, new_room)
                
            else:
                new_block = random.choice(time_blocks)
                new_room = random.choice(list(rooms.keys()))
                individual[i] = (new_block, new_room)
    
    return individual

# Probar tu operador
original = create_random_individual()
print("Original:")
print_schedule(original)
print(f"Fitness: {fitness(original)}")

mutated = [x for x in original]  # Copia
mutated = mutate(mutated, 0.5)  # 50% para ver más mutaciones

print("\nMutado (prob=50%):")
print_schedule(mutated)
print(f"Fitness: {fitness(mutated)}")
````

# Función de Evolución (Proporcionada)

Esta función combina todos tus operadores:

````python
def selection(population):
    """Selección por torneo de 3 individuos."""
    tournament = random.sample(population, 3)
    return max(tournament, key=fitness)

def evolve(population, generations=50, verbose=True):
    best_overall = None
    best_overall_fitness = float('-inf')
    
    fitness_history = []   # Guardamos la evolución
    
    for gen in range(generations):
        
        # Aplicamos elitismo, mantenemos los 2 mejores
        sorted_pop = sorted(population, key=fitness, reverse=True)
        new_population = sorted_pop[:2]
        
        #generamos el resto de la población
        while len(new_population) < len(population):
            parent1 = selection(population)
            parent2 = selection(population)
            child = crossover(parent1, parent2)
            child = mutate(child)
            new_population.append(child)
        
        population = new_population
        
        #buscamos el mejor de esta generación
        best = max(population, key=fitness)
        best_fit = fitness(best)
        
        fitness_history.append(best_fit)  # Guardamos el historial
        
        if best_fit > best_overall_fitness:
            best_overall = best
            best_overall_fitness = best_fit
        
        if verbose and (gen % 10 == 0 or gen == generations - 1):
            print(f"Gen {gen+1:3d}: Mejor fitness = {best_fit:6.1f}")
    
    return best_overall, fitness_history  # retornamos el mejor fitness de la generación y el historial de los mejores fitness

print("✓ Funciones de evolución cargadas")
````

# Ejecutar el Algoritmo Genético

Una vez completadas las tres tareas, ejecuta esta celda:

````python
print("="*70)
print("INICIANDO ALGORITMO GENÉTICO")
print("="*70)

# Crear población inicial
initial_pop = create_population(size=30)

print(f"\nPoblación inicial: {len(initial_pop)} individuos")
print(f"Fitness promedio: {sum(fitness(ind) for ind in initial_pop) / len(initial_pop):.1f}")
print(f"Mejor fitness: {max(fitness(ind) for ind in initial_pop):.1f}")

print("\n" + "="*70)
print("EVOLUCIÓN")
print("="*70)

# Ejecutar
best_solution, history = evolve(initial_pop, generations=100)

print("\n" + "="*70)
print("MEJOR SOLUCIÓN")
print("="*70)
print_schedule(best_solution)
print(f"\nFitness final: {fitness(best_solution):.1f}")

# Análisis
print("\n" + "="*70)
print("ANÁLISIS")
print("="*70)

# Conflictos
conflicts = 0
for i in range(len(best_solution)):
    for j in range(i+1, len(best_solution)):
        if best_solution[i][0] == best_solution[j][0] and best_solution[i][1] == best_solution[j][1]:
            conflicts += 1

# Capacidad
capacity_violations = 0
for i, (block, room) in enumerate(best_solution):
    if courses[i][1] > rooms[room]:
        capacity_violations += 1

# Profesores
professor_conflicts = 0
for c1, c2 in same_professor:
    if best_solution[c1][0] == best_solution[c2][0]:
        professor_conflicts += 1

print(f"Conflictos de salón: {conflicts}")
print(f"Violaciones de capacidad: {capacity_violations}")
print(f"Conflictos de profesor: {professor_conflicts}")

if conflicts == 0 and capacity_violations == 0 and professor_conflicts == 0:
    print("\n✓ ¡SOLUCIÓN VÁLIDA! Todas las restricciones duras satisfechas.")
else:
    print("\n✗ Solución inválida. Hay restricciones violadas.")
````

# Justificación de Decisiones de Diseño

## 1. Parámetros del Algoritmo Genético

### Tamaño de Población: 30 individuos
**Justificación:**
- Poblaciones pequeñas (10): convergen rápido pero caen en óptimos locales
- Poblaciones grandes (50): más lentas sin mejora significativa en calidad
- **30 individuos** ofrece el mejor equilibrio entre diversidad genética y tiempo de ejecución
- Confirmado por el experimento de comparación de poblaciones

### Número de Generaciones: 100
**Justificación:**
- Observamos que el fitness se estabiliza entre las generaciones 10-20
- 100 generaciones aseguran convergencia completa
- Permite observar mejoras incrementales en restricciones blandas
- Tiempo de ejecución razonable (<5 segundos)

### Tasa de Mutación: 20% (prob=0.2)
**Justificación:**
- Tasas muy altas (>50%) destruyen buenas soluciones (búsqueda aleatoria)
- Tasas muy bajas (<5%) reducen exploración
- **20%** mantiene diversidad sin destruir convergencia
- Permite escape de óptimos locales

---

## 2. Operadores Genéticos

### Selección: Torneo de 3 individuos
**Justificación:**
- Más simple que selección por ruleta
- Mantiene presión selectiva moderada
- Conserva diversidad (no solo el mejor se reproduce)
- Tamaño 3 es estándar y eficiente

### Cruce: Un punto (Original) / Dos puntos (Bonus)
**Justificación:**
- **Un punto:** Simple, probado, efectivo para este problema
- **Dos puntos (Bonus):** Mayor mezcla genética, mejor para evitar bloques de genes "pegados"
- Ambos preservan partes buenas de los padres

### Mutación: Inteligente (Bonus)
**Justificación Original (Aleatoria):**
- Introduce variación aleatoria necesaria
- Explora diferentes bloques y salones
- Mantiene población dinámica

**Justificación Bonus (Inteligente):**
- **Dirigida:** corrige violaciones específicas detectadas
- **Más eficiente:** no hace cambios aleatorios innecesarios
- **Convergencia más rápida:** soluciones válidas en menos generaciones
- **Criterio de prioridad:**
  1. Conflictos de salón (crítico)
  2. Capacidad insuficiente (crítico)
  3. Profesores compartidos (crítico)

### Elitismo: 2 mejores individuos
**Justificación:**
- Garantiza que las mejores soluciones no se pierdan
- 2 individuos (6.7% de 30) es suficiente sin reducir diversidad
- Acelera convergencia manteniendo exploración

---

## 3. Función de Fitness

### Puntaje Base: 200 puntos
**Justificación:**
- Suficientemente alto para permitir diferenciación
- Permite penalizaciones significativas sin llegar a negativo

### Penalizaciones Duras: -50 puntos cada una
**Justificación:**
- **Diferencia significativa** vs restricciones blandas (-3 a -5)
- Garantiza que **soluciones válidas > soluciones inválidas**
- Con 3 tipos de restricciones duras, máximo -150 por cada tipo
- Prioriza corrección sobre optimización

### Penalizaciones Blandas: -3 a -5 puntos
**Justificación:**
- **Diferenciación:** -5 para bloques tardíos (más importante) vs -3 desperdicio
- Suficiente para guiar búsqueda sin dominar restricciones duras
- Permite comparar soluciones válidas entre sí

## Opción 4: Nueva Restricción
Añade penalización si dos cursos relacionados están muy cerca (mismo día).

````python
# Opción 4 - Nueva Restricción: Penalización por cursos relacionados en el mismo día

# Mapeo de bloques a días (para identificar cursos "muy cerca")
block_to_day = {
    1: 'Lunes', 2: 'Lunes',           # Lunes 8-10, 10-12
    3: 'Miércoles', 4: 'Miércoles',   # Miércoles 8-10, 10-12
    5: 'Viernes'                       # Viernes 8-10
}

def fitness_con_restriccion_cercania(individual):
    """
    Añade penalización si dos cursos RELACIONADOS (mismo profesor) 
    están MUY CERCA (mismo día, aunque sean bloques diferentes).
    
    Restricción BLANDA adicional:
    - -2 puntos si cursos del mismo profesor están en el mismo día
    """
    score = fitness(individual)  # Usar fitness base
    
    # Penalizar cursos relacionados programados muy cerca (mismo día)
    for c1, c2 in same_professor:
        bloque1, bloque2 = individual[c1][0], individual[c2][0]
        dia1, dia2 = block_to_day[bloque1], block_to_day[bloque2]
        
        # Si están en el mismo día pero diferente bloque → muy cerca
        if bloque1 != bloque2 and dia1 == dia2:
            score -= 2  # Penalización por proximidad
            
    return score

print("-" * 60)
print("OPCIÓN 4: NUEVA RESTRICCIÓN DE PROXIMIDAD")
print("-" * 60)

# Prueba
individuo = create_random_individual()
print_schedule(individuo)

# Comparar fitness
print(f"\nFitness estándar:        {fitness(individuo)}")
print(f"Fitness con proximidad:  {fitness_con_restriccion_cercania(individuo)}")
print(f"Diferencia (penalización): {fitness(individuo) - fitness_con_restriccion_cercania(individuo)}")

# Verificar cursos relacionados
print("\nAnálisis de cursos relacionados (mismo profesor):")
print("-" * 60)
for c1, c2 in same_professor:
    b1, b2 = individuo[c1][0], individuo[c2][0]
    n1, n2 = courses[c1][0], courses[c2][0]
    d1, d2 = block_to_day[b1], block_to_day[b2]
    
    print(f"{n1} vs {n2}:")
    print(f"  {n1}: {block_names[b1]} ({d1})")
    print(f"  {n2}: {block_names[b2]} ({d2})")
    
    if b1 == b2:
        print(f"  → ❌ MISMO BLOQUE (restricción dura violada, -50 pts)")
    elif d1 == d2:
        print(f"  → ⚠️  MISMO DÍA pero diferente bloque (-2 pts por proximidad)")
    else:
        print(f"  → ✓ Días diferentes (sin penalización adicional)")
    print()
````
